# Cell1:setup

# OUTPUTS

```python
import sys
sys.path.append("..")

from src.agent import get_llm, build_agent
from src.planner import plan_trip

llm = get_llm()
agent = build_agent()
print("Ready")
```

# Cell2:plan a trip

```python
query = (
    "Plan a 3-day trip from Bangalore to Goa starting 2026-11-20. "
    "I like beaches and want a hotel with at least 4 stars."
)
result = plan_trip(agent, llm, query)

if result.get("error"):
    print("ERROR:", result["error"])
else:
    print(result["text"])
```

# Cell3:JSON outputs

```python
if not result.get("error"):
    print(result["json"])
```

# Cell4:save outputs

```python
from pathlib import Path

out_dir = Path("../output")
out_dir.mkdir(exist_ok=True)
(out_dir / "trip_plan.json").write_text(result["json"], encoding="utf-8")
(out_dir / "trip_plan.txt").write_text(result["text"], encoding="utf-8")
print("Saved to", out_dir.resolve())
```

# Cell5:edge cases

```python
for q in ["Plan a 3-day trip from Delhi to Goa starting 2026-11-20.",
          "Plan me a nice trip somewhere warm."]:
    r = plan_trip(agent, llm, q)
    print("Q:", q)
    print(r.get("error") or r["text"])
    print("-" * 60)
```
