# DATA EXPLORATION

# Cell1:load the files

```python
import json
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data")

def load_json(filename):
    """Load a JSON file from the data directory."""
    with open(DATA_DIR / filename, "r", encoding="utf-8") as f:
        return json.load(f)

flights_raw = load_json("flights.json")
hotels_raw = load_json("hotels.json")
places_raw = load_json("places.json")

for name, data in [("flights", flights_raw), ("hotels", hotels_raw), ("places", places_raw)]:
    print(name, "->", type(data).__name__, "| size:", len(data))
```

# Cell2: look the structure 

```python
def peek(name, data):
    print(f"\n===== {name} =====")
    if isinstance(data, list):
        print("First record:")
        print(json.dumps(data[0], indent=2, ensure_ascii=False))
    elif isinstance(data, dict):
        print("Top-level keys:", list(data.keys()))
        first_key = list(data.keys())[0]
        print(f"Sample under '{first_key}':")
        print(json.dumps(data[first_key], indent=2, ensure_ascii=False)[:1200])

peek("flights", flights_raw)
peek("hotels", hotels_raw)
peek("places", places_raw)
```

# Cell3: dataframes and quality check

```python
def to_df(data):
    """Convert list or dict-of-lists JSON into a DataFrame."""
    if isinstance(data, dict):
        for value in data.values():
            if isinstance(value, list):
                return pd.DataFrame(value)
    return pd.DataFrame(data)

flights_df, hotels_df, places_df = to_df(flights_raw), to_df(hotels_raw), to_df(places_raw)

for name, df in [("flights", flights_df), ("hotels", hotels_df), ("places", places_df)]:
    print(f"\n===== {name} ({df.shape[0]} rows, {df.shape[1]} cols) =====")
    print(df.dtypes)
    print("Missing values:\n", df.isna().sum())
    display(df.head())
```
