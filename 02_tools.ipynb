```python
import sys
import json
sys.path.append("..")   # lets the notebook import from src/

from src.tools import (
    search_flights, recommend_hotels, discover_places,
    get_weather, estimate_budget, ALL_TOOLS,
)

def show(result):
    print(json.dumps(result, indent=2, ensure_ascii=False))

print([t.name for t in ALL_TOOLS])
```

```python
print("--- Cheapest Delhi -> Kolkata ---")
show(search_flights.invoke({"source": "Delhi", "destination": "Kolkata", "sort_by": "cheapest"}))

print("--- Fastest Bangalore -> Goa ---")
show(search_flights.invoke({"source": "Bangalore", "destination": "Goa", "sort_by": "fastest"}))

print("--- Route that does not exist: Delhi -> Goa ---")
show(search_flights.invoke({"source": "Delhi", "destination": "Goa"}))
```
